```python
order_analysis = orders.merge(
    customers,
    on="customer_id",
    how="left"
)
```

```python
order_analysis = order_analysis.merge(
    order_items,
    on="order_id",
    how="left"
)
```

```python
order_analysis = order_analysis.merge(
    products,
    on="product_id",
    how="left"
)
```

```python
order_analysis = order_analysis.merge(
    sellers,
    on="seller_id",
    how="left"
)
```

```python
order_analysis = order_analysis.merge(
    category_translation,
    on="product_category_name",
    how="left"
)
```

```python
order_analysis["order_value"] = (
    order_analysis["price"] +
    order_analysis["freight_value"]
)
```

```python
order_analysis["delivery_days"] = (
    order_analysis["order_delivered_customer_date"]
    - order_analysis["order_purchase_timestamp"]
).dt.days
```

```python
order_analysis["estimated_delivery_days"] = (
    order_analysis["order_estimated_delivery_date"]
    - order_analysis["order_purchase_timestamp"]
).dt.days
```

```python
order_analysis["delivery_delay"] = (
    order_analysis["order_delivered_customer_date"]
    - order_analysis["order_estimated_delivery_date"]
).dt.days
```

```python
order_analysis["delivery_status"] = np.where(
    order_analysis["delivery_delay"].isna(),
    "Not Delivered",
    np.where(
        order_analysis["delivery_delay"] > 0,
        "Late",
        "On Time"
    )
)
```

```python
order_analysis["order_year"] = (
    order_analysis["order_purchase_timestamp"].dt.year
)

order_analysis["order_month"] = (
    order_analysis["order_purchase_timestamp"].dt.month
)

order_analysis["order_quarter"] = (
    order_analysis["order_purchase_timestamp"].dt.quarter
)

order_analysis["order_month_name"] = (
    order_analysis["order_purchase_timestamp"].dt.strftime("%b")
)
```

```python
order_analysis["freight_percentage"] = np.where(
    order_analysis["price"] > 0,
    order_analysis["freight_value"] / order_analysis["price"] * 100,
    np.nan
)
```

```python
order_analysis["product_category"] = (
    order_analysis["product_category_name_english"]
    .fillna(order_analysis["product_category_name"])
)
```
